```python
import pandas as pd
import numpy as np
from datetime import timedelta

np.random.seed(42)
n = 500

# تاریخ‌ها
hire_dates = [pd.Timestamp('2018-01-01') + timedelta(days=np.random.randint(0, 2000)) for _ in range(n)]
statuses = np.random.choice(['Active', 'Terminated'], n, p=[0.8, 0.2])

# تاریخ خروج: فقط برای Terminated
termination_dates = []
for status, hire in zip(statuses, hire_dates):
    if status == 'Terminated':
        termination_dates.append(hire + timedelta(days=np.random.randint(30, 1000)))
    else:
        termination_dates.append(pd.NaT)  # NaT = Not a Time (خالی برای تاریخ)

df_emp = pd.DataFrame({
    'EmployeeID': range(1001, 1001 + n),
    'Name': [f'Employee_{i}' for i in range(n)],
    'DeptID': np.random.choice([1, 2, 3, 4, 5], n),
    'HireDate': hire_dates,
    'Status': statuses,
    'Salary': np.random.randint(40000, 150000, n),
    'PerformanceScore': np.random.randint(1, 6, n),
    'JobSatisfaction': np.random.randint(1, 5, n),
    'TerminationDate': termination_dates
})

df_emp.to_csv('hr_employees.csv', index=False)
print("hr_employees.csv ساخته شد")
print(df_emp['Status'].value_counts())

import pandas as pd

# جدول Departments
df_depts = pd.DataFrame({
    'DeptID': [1, 2, 3, 4, 5],
    'DeptName': ['Sales', 'IT', 'HR', 'Marketing', 'Finance'],
    'Manager': ['Alice', 'Bob', 'Charlie', 'David', 'Eva']
})
df_depts.to_csv('hr_departments.csv', index=False)

# جدول Calendar
date_range = pd.date_range(start='2018-01-01', end='2025-01-01')
df_cal = pd.DataFrame({
    'Date': date_range,
    'Year': date_range.year,
    'Month': date_range.month,
    'MonthName': date_range.strftime('%B'),
    'Quarter': date_range.quarter
})
df_cal.to_csv('calendar.csv', index=False)

print("hr_departments.csv ساخته شد")
print("calendar.csv ساخته شد")

```
